```python
import pyvista as pv
import subprocess
```

```python
res = subprocess.run(["umsh", "msh/msh.tel", "msh/msh.cgns"], capture_output=True)

print(str(res.stdout))

reader = pv.CGNSReader('msh/msh.cgns')
msh = reader.read()
plotter = pv.Plotter()

plotter.add_mesh(msh, style='wireframe')
plotter.camera_position = 'xy'
plotter.show_axes()
plotter.show(jupyter_backend='static')
```
